```python
import numpy as np
import pandas as pd
```

```python

# CLEAN DATA
# list of desired years
years = ["2017", "2019" , "2021", "2023"]

# list of desired variables

## person level
desired_col = ["CONTROL","AGE", "CITSHP", "ENROLL", "FAMLINE", "FAMREL", "GRAD", "FAMTYP", "MAR", "MIL", "MOVE", 
               "MOVERGRP", "PERSONID", "RACE", "RESPONDENT", "SEX", "SPAN"]
## household level
desired_col_hh = ["OMB13CBSA","CONTROL", "INTSTATUS", "MGRONSITE", "TENURE", "BLD", "BATHROOMS", "BEDROOMS", "UNITSIZE", "WASHER", "MOLDBEDRM", 
               "MOLDKITCH","MOLDLROOM", "ROACH", "WALLCRACK", "GRANDHH", "HSHLDTYPE", "MILHH", "MULTIGEN", "NUMPEOPLE",
               "SAMEHHLD", "HHAGE", "HHENROLL", "HHGRAD", "HHCITSHP","HHMAR", "HHRACE", "HHSEX", "HHSPAN", "FINCP", "RENT", "TOTHCAMT",
               "HUDSUB", "RENTCNTRL", "RENTSUB", "NEARABAND", "NEARBARCL", "NEARTRASH", "NHQPCRIME", "NHQPUBTRN", "NHQSCRIME",
              "RATINGHS", "RATINGNH", "HRATE", "NRATE", "SEARCHFAM", "SEARCHLIST", "SEARCHNET", "SEARCHOTH", "SEARCHPUB", "SEARCHNET",
              "SEARCHOTH", "SEARCHPUB", "SEARCHREA", "SEARCHSIGN", "MOVFORCE", "MOVWHY", "RMCHANGE", "RMCOMMUTE", "RMCOSTS", "RMFAMILY",
              "RMHOME", "RMHOOD", "RMJOB", "RMOTHER", "RMOWNHH"]

# loop through each data file
for i in years:
    # reading and cleaning person-level data
    data = pd.read_csv(f"AHS_{i}_metro/person.csv")
    data = data[desired_col]
    for u in range(len(data.columns)):
        if data.iloc[:,u].dtypes == 'O':
            data.iloc[:,u] = data.iloc[:,u].str.replace("'", "")
    # reading and cleanining household-level data
    hhold = pd.read_csv(f"AHS_{i}_metro/household.csv")
    hhold = hhold[desired_col_hh]
    for u in range(len(hhold.columns)):
        if hhold.iloc[:,u].dtypes == 'O':
            hhold.iloc[:,u] = hhold.iloc[:,u].str.replace("'", "")

    # merge
    merged = pd.merge(data, hhold, on = 'CONTROL', how = 'left')

    # save as csv
    merged.to_csv(f"AHS_{i}_merged_cleaned.csv", index=False)
                       
```

```python
# create fully cleaned data set
ahs2017 = pd.read_csv("AHS_2017_merged_cleaned.csv")
ahs2019 = pd.read_csv("AHS_2019_merged_cleaned.csv")
ahs2021 = pd.read_csv("AHS_2021_merged_cleaned.csv")
ahs2023 = pd.read_csv("AHS_2023_merged_cleaned.csv")

# add year variable
ahs2017["year"] = 2017
ahs2019["year"] = 2019
ahs2021["year"] = 2021
ahs2023["year"] = 2023

```

```python
# create final data frame
final_df = pd.concat([ahs2017, ahs2019, ahs2021, ahs2023], axis = 0)
```

```python
final_df = final_df.replace([-6, -9], np.nan)
```

```python
# export to csv
final_df.to_csv("AHS_full_data.csv", index=False)
```
